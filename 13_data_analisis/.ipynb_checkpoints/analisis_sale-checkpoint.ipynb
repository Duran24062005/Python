```python
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
df['editora'] = df['editora'].fillna('Desconocida')
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['anio'])
```

```python
df.info()
```

```python
df['anio']
```

```python
df['anio'] = df['anio'].astype(int)
```

```python
df.head()
```

```python
df=df.reset_index(drop=True)
```

```python
ventas_por_genero = df.groupby('genero')['ventas_global'].sum().sort_values(ascending=False)
```

```python
ventas_por_genero.plot(kind='bar')
```

```python
ventas_por_genero.plot(kind='bar')
plt.title('Ventas globales por genero')
plt.ylabel('Millones de copias')
plt.xlabel('Genero')
plt.show()
```

```python
ventas_por_genero.plot(kind='barh')
plt.title('Ventas globales por genero')
plt.xlabel('Millones de copias')
plt.ylabel('Genero')
plt.show()
```

```python
ventas_por_genero.plot(kind='pie')
plt.title('Ventas globales por genero')
plt.show()
```

```python
ventas_por_genero.plot(kind='line')
plt.title('Ventas globales por genero')
plt.ylabel('Millones de copias')
plt.xlabel('Genero')
plt.show()
```

```python
ventas_por_annio = df.groupby('anio')['ventas_global'].sum()
```

```python
ventas_por_annio.plot() # Por defecto plot genera un grafico de linea
plt.title('Ventas por año')
plt.ylabel('Millones de Copias')
plt.xlabel('Años')
plt.show()
```
